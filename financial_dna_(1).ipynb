```python
import lightgbm as lgb
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, root_mean_squared_error
from sklearn.model_selection import train_test_split

# 1. Feature Selection & Target Definition
# Drop target leakage columns (balances, future payment dates, IDs)
leakage_cols = [
    "Account Holder Name",
    "Account ID",
    "Original Account ID",
    "Principal Balance",
    "Disbursed Amount",
    "Approval Date",
    "Activation Date",
    "First Repayment Date",
    "Last Payment Date",
    "Locked Date",
]

# Define target and initial feature pool based on your dataframe peek
target = "Loan Amount"
features = [
    col for col in df.columns if col not in leakage_cols and col != target
]

X = df[features].copy()
y = df[target].copy()
```

```python
# 2. Handle Categorical Features for LightGBM
categorical_cols = [
    "Branch",
    "Product",
    "Loan usage Business",
    "Environmental & Social Risk",
    "Age Band (Client)",
    "Nature of business",
]

# Ensure they exist in features and convert to category type
actual_cat_cols = [col for col in categorical_cols if col in X.columns]
for col in actual_cat_cols:
    X[col] = X[col].astype("category")
```

```python
# 3. Log-Transform Target to Handle Financial Skewness
y_log = np.log1p(y)

```

```python
# 4. Train-Test Split
X_train, X_test, y_train_log, y_test_log = train_test_split(
    X, y_log, test_size=0.2, random_state=42
)
```

```python
# 5. Initialize LightGBM Regressor
# Reuses tree structures but shifts objective to continuous optimization
reg_model = lgb.LGBMRegressor(
    objective="regression",  # Regression mode
    metric="rmse",  # Optimizing root mean squared error
    n_estimators=1000,
    learning_rate=0.03,
    num_leaves=31,
    random_state=42,
    verbose=-1,
)
```

```python
#  Convert Categorical / Object columns to Category dtype
# LightGBM can handle the 'category' type automatically if you specify it
categorical_cols = [
    'Account State', 'Account Sub-State', 'Gender (Client)',
    'Loan Usage', 'Was Refinanced', 'Was Rescheduled', 'Risk Level'
]
for col in categorical_cols:
    X_train[col] = X_train[col].astype('category')
    X_test[col] = X_test[col].astype('category')  # Don't forget your test set!

#  Extract numeric features from your DateTime columns
date_cols = ['Expected Maturity Date', 'Created (Client)']
for col in date_cols:
    # Convert to datetime first just in case
    X_train[col] = pd.to_datetime(X_train[col])
    X_test[col] = pd.to_datetime(X_test[col])

    # Extract numerical components (e.g., Year, Month, Day)
    X_train[f'{col}_year'] = X_train[col].dt.year
    X_train[f'{col}_month'] = X_train[col].dt.month
    X_train[f'{col}_day'] = X_train[col].dt.day

    X_test[f'{col}_year'] = X_test[col].dt.year
    X_test[f'{col}_month'] = X_test[col].dt.month
    X_test[f'{col}_day'] = X_test[col].dt.day

#  Drop the original raw datetime columns so LightGBM doesn't reject them
X_train = X_train.drop(columns=date_cols)
X_test = X_test.drop(columns=date_cols)
```

```python
# 6. Fit Model with Early Stopping
reg_model.fit(
    X_train,
    y_train_log,
    eval_set=[(X_test, y_test_log)],
    categorical_feature=categorical_cols, # Tell LightGBM which columns are categories
    callbacks=[lgb.early_stopping(stopping_rounds=30, verbose=False)],
)
```

```python
# 7. Predict & Inverse-Transform Predictions back to Currency
y_pred_log = reg_model.predict(X_test)
y_pred = np.expm1(y_pred_log)
y_test_actual = np.expm1(y_test_log)

```

```python
# 8. Post-Processing & Financial Guardrails
# Enforce floor limits ($500 equivalent) and round to the nearest 100 units
final_limits = np.maximum(y_pred, 500)
final_limits = np.round(final_limits / 100) * 100
```

```python
# 9. Evaluate Performance
mae = mean_absolute_error(y_test_actual, final_limits)
rmse = root_mean_squared_error(y_test_actual, final_limits)

print(f"Regression Engine Evaluation Metrics:")
print(f"• Mean Absolute Error (MAE): {mae:,.2f} Units")
print(f"• Root Mean Squared Error (RMSE): {rmse:,.2f} Units")
```
